```python
#compare two car feature set
set1={'a','b','c','d'}
set2={'a','e','c','k'}
```

```python
#common feature in both car->intersection
common_feature=set1.intersection(set2)
print(common_feature)
```

```python
#union
all_feature=set1.union(set2)
print(all_feature)
```

```python
#features that are in set2 but not in set 1
diff=set2.difference(set1)
print(diff)
```
