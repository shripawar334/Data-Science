## 📊 Model Evaluation Summary Table  

| **Metric** | **What it Measures**               | **Ideal Value** | **Use Case** |
|------------|------------------------------------|-----------------|--------------|
| **R²**     | % variance explained               | Closer to 1     | Model fit quality |
| **RMSE**   | Average squared error              | Closer to 0     | Penalizes large errors |
| **MAE**    | Average absolute error             | Closer to 0     | Simpler error interpretation |
| **MAPE**   | Avg % error from actuals           | Lower %         | Explaining % error to audience |
| **Adjusted R²** | Adjusted variance explained   | Closer to 1     | Better measure when multiple predictors are used |

---

### ✅ Key Points:
- **R²** shows how much variation in the target is explained by the model.  
- **Adjusted R²** is more reliable when you have multiple independent variables.  
- **RMSE** (Root Mean Squared Error) punishes large errors more heavily.  
- **MAE** (Mean Absolute Error) is easier to interpret because it's in the same unit as the target.  
- **MAPE** (Mean Absolute Percentage Error) is great when you want to explain error in % terms to a non-technical audience.  



## ✅ Final Notes:

- Always evaluate the model on **test data**  
- Use **multiple metrics** to get a complete picture  
- **RMSE** penalizes large errors more; **MAE** is easier to explain  
- **R²** shows explanatory power, not accuracy  

---

Use these metrics to select and justify your regression model before making business decisions or deployment.


## 1.r2_score

```python
from sklearn.metrics import r2_score
```

```python
actual=[10,15,15]
pred=[12,14,16]
print('R2_score:',r2_score(actual,pred))
```

## 2.RMSE (root mean square error)

```python
from sklearn.metrics import mean_squared_error
import numpy as np
```

```python
actual=[10,15,15]
pred=[12,14,16]
rmse=np.sqrt(mean_squared_error(actual,pred))
print('RMSE:',rmse)
```

## 3.MAE (mean absolute error)

```python
from sklearn.metrics import mean_absolute_error
```

```python
actual=[10,15,15]
pred=[12,14,16]
mae=mean_absolute_error(actual,pred)
print('MAE:',mae)
```

## 4.MAPE (mean absolute percentage error)

```python
actual=np.array([10,15,15])
pred=np.array([12,14,16])
mape=np.mean(np.abs((actual-pred)/actual))*100
print('MAPE:',mape)
```
