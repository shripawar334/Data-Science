# 📈 Predicting Sales Based on Store Footfalls using Simple Linear Regression

We want to predict store sales based on footfall using a simple linear regression model.  

Just like more people in a store usually means more sales, we’ll use this pattern to predict future sales.  

---

## 🔹 Linear Regression Overview  

Linear Regression is a supervised learning algorithm used to model the relationship between one or more independent variables (features) and a continuous dependent variable (target).  

### Types of Linear Regression:  

**1. Simple (or Single) Linear Regression:**  
Uses one independent variable to predict a continuous target variable by fitting a straight line.  

**2. Multiple Linear Regression:**  
Uses two or more independent variables to predict the target by fitting a multi-dimensional hyperplane.  


```python
import pandas as pd
import matplotlib.pyplot as plt 
from sklearn.linear_model import LinearRegression
```

```python
data=pd.read_csv('store_sales.csv')
data.head()
```

## define independent variable(footfalls) and depedent variable (dailysales)

```python
# ✅ Independent variable (X) → Footfalls
# Using double brackets [['Footfalls']] to keep X as a 2D DataFrame (shape = n_samples x 1)
# ML models like LinearRegression always expect X in 2D format
X = data[['Footfalls']]  

# ✅ Dependent variable (Y) → Daily Sales
# Here we use single brackets ['DailySales'] to keep Y as a 1D Series (target vector)
Y = data['DailySales']
```

```python
type(X)
```

```python
type(Y)
```

## visualize data and check relationship between X and Y 

```python
plt.xlabel('footfalls')
plt.ylabel('daily_sales')
plt.scatter(X,Y,color='red')
plt.show()
```

```python
from sklearn.linear_model import LinearRegression
```

This line imports the **LinearRegression** class from scikit-learn, a popular machine learning library.  

It allows us to create a simple linear regression model,  
which finds the best-fit straight line to predict a target variable based on input features.  


## Create Train linear regression model

```python
model=LinearRegression()
model.fit(X,Y)
```

## Print the model parameter


```python
print(f"Intercept:{model.intercept_}")
print(f"coefficient(slope):{model.coef_}")
```

### simple linear regression equation y = β0 + β1 * x



y = Predicted value (dependent variable, e.g., sales)  

x = Input value (independent variable, e.g., footfall)  

β1 = Slope (coefficient showing how much y changes with x)  

β0 = Intercept (value of y when x = 0)  



```python
model.predict([[20]])
```

```python
model.predict([[35]])
```

```python
new_footfall=pd.DataFrame({
    'Footfalls':[232,33,65,34]
})
new_footfall
```

```python
model.predict(new_footfall)
```
