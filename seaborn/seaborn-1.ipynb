# Seaborn

## 📊 Quick Comparison

| Feature        | Matplotlib                       | Seaborn                         |
|----------------|----------------------------------|---------------------------------|
| **Level**      | Low-level (more control)         | High-level (easier to use)      |
| **Appearance** | Basic                            | Styled by default               |
| **Plot Types** | Basic charts                     | Statistical charts (box, violin)|
| **Integration**| Works with all data              | Best with Pandas DataFrames     |

| **Learning Curve** | Moderate                     | Easier for beginners            |


```python
import pandas as pd 
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('network_data_large.csv')
```

```python
# Create the scatter plot
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="Call_Duration", y="Satisfaction_Score", hue="Network_Provider")

# Move the legend outside the plot
plt.legend(title="Network Provider", bbox_to_anchor=(1, 1), loc="upper left")

# Add Labels and title
plt.xlabel("Call Duration (minutes)")
plt.ylabel("Satisfaction Score")
plt.title("Call Duration vs. Satisfaction Score")

# Show the plot
plt.show()
```

# Categorical Plot: Count of Customers by Network Provider

## 📊 Categorical Plots

**Definition:**  
Categorical plots display relationships involving **categorical variables** (e.g., gender, department, product category).

---

**Usage:**  
- Used to analyze **grouped data** (e.g., average number of calls per customer segment).  
- Helps in comparing different categories.  

---

**Characteristics:**  
- Includes **bar plots** (`sns.barplot`), **box plots** (`sns.boxplot`), **violin plots** (`sns.violinplot`), etc.  
- Allows **comparison across multiple groups**.  

---

**Applications:**  
- **Telecommunication:** Distribution of **customer complaints by network provider**.  
- **Healthcare:** Distribution of **patient recovery time by disease**.  
- **Education:** Comparison of **students' performance by subject**.  


```python
# Categorical Plot: Count of Customers by Network Provider
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='Network_Provider', hue='Network_Provider', palette='Set2')

plt.title("Customer Distribution by Network Provider")
plt.xlabel("Network Provider")
plt.ylabel("Count of Customers")

# plt.legend().remove()   # ✅ This removes the Legend
plt.show()
```

#  Distribution Plots

## 📊 Distribution Plots

**Definition:**  
Distribution plots visualize the **spread and shape** of data.  

---

**Usage:**  
- Helps understand **data distribution**, skewness, and outliers.  
- Useful for **identifying trends and anomalies**.  

---

**Characteristics:**  
- Includes **histograms** (`sns.histplot`), **kde plots** (`sns.kdeplot`), and **rug plots** (`sns.rugplot`).  
- Displays **density estimates** and **probability distributions**.  

---

**Applications:**  
- **Telecommunication:** Distribution of **call durations per customer**.  
- **Finance:** Distribution of **transaction amounts in a bank**.  
- **E-commerce:** Analyzing **purchase frequency of products**.  


# Distribution Plot: Distribution of Call Durations

```python
# Distribution Plot: Distribution of Call Durations
plt.figure(figsize=(7, 4))
sns.histplot(df['Call_Duration'], bins=20, kde=True, color='blue')

plt.title("Distribution of Call Durations")
plt.xlabel("Call Duration (Minutes)")
plt.ylabel("Frequency")
plt.show()
```

## 🔥 Matrix Plots

## 🔥 Matrix Plots

**Definition:**  
Matrix plots visualize **relationships between multiple variables** using **heatmaps** and **clustering methods**.  

---

**Usage:**  
- Helps find **correlations** and **patterns** in datasets.  
- Used in **feature selection** and **anomaly detection**.  

---

**Characteristics:**  
- Includes **heatmaps** (`sns.heatmap`) and **clustermaps** (`sns.clustermap`).  
- Displays **intensity** and **similarity** between variables.  

---

**Applications:**  
- **Telecommunication:** Heatmap of **network congestion across different locations**.  
- **Finance:** Correlation between **various stock indices**.  
- **Weather Analytics:** Heatmap of **temperature variations over time**.  


```python
plt.figure(figsize=(8, 6))

# Compute correlation for numeric columns only
numeric_df = df.select_dtypes(include=['number'])   # Select only numeric columns

sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)

plt.title("Correlation Matrix of Features")
plt.show()
```

# Box Plot: Satisfaction Score by Network Provider

```python
# Box Plot: Satisfaction Score by Network Provider
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='Network_Provider', y='Satisfaction_Score')

plt.xlabel("Network Provider")
plt.ylabel("Satisfaction Score")
plt.title("Satisfaction Score Across Different Network Providers")
plt.show()
```

# Violin Plot: Call Duration by Network Provider

### Swarmplots are useful for displaying individual data points, especially when dealing with categorical variables.


```python
# Violin Plot: Call Duration by Network Provider
plt.figure(figsize=(8, 5))
sns.violinplot(data=df, x='Network_Provider', y='Call_Duration')

plt.xlabel("Network Provider")
plt.ylabel("Call Duration (minutes)")
plt.title("Call Duration Distribution Across Network Providers")
plt.show()
```
